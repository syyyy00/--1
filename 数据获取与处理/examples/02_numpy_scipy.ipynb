```python
# ==============================
# 02 数值计算练习
# 解方程、拟合、积分、优化
# ==============================

import numpy as np
from numeric_utils import solve_linear, poly_fit, numerical_integral

# --------------------
# 1. 解线性方程组
# --------------------
A = np.array([[1,2],[3,4]])
b = np.array([5,6])
x = solve_linear(A, b)
print("Ax=b 的解：", x)

# --------------------
# 2. 多项式拟合
# --------------------
x_data = [1,2,3,4,5]
y_data = [1,4,9,16,25]
coef = poly_fit(x_data, y_data, 2)
print("拟合系数（高次→低次）：", coef)

# --------------------
# 3. 数值积分
# --------------------
def f(x):
    return x**2

res = numerical_integral(f, 0, 2)
print("积分结果：", res)

# --------------------
# 4. SciPy 优化（求最小值）
# --------------------
from scipy.optimize import minimize
def func(x):
    return x**2 + 4*x + 5

ret = minimize(func, x0=0)
print("最小值点：", ret.x)
print("最小值：", ret.fun)
```
