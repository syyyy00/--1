```python
# ==============================
# 04 绘图大全练习
# 论文所有图表都在这里
# ==============================

import numpy as np
from plot_library import plot_lines, scatter_plot, heatmap, plot_3d_surface

# --------------------
# 1. 折线图（负荷/时序）
# --------------------
x = np.linspace(0,24,96)
y1 = np.sin(x/12*np.pi)*50 + 100
y2 = y1 + np.random.randn(len(y1))*2

plot_lines(
    x=x,
    y_dict={"真实":y1, "预测":y2},
    xlabel="时间", ylabel="负荷",
    title="负荷曲线对比",
    save_path="负荷图.png"
)

# --------------------
# 2. 散点图（相关性）
# --------------------
x = np.random.randn(100)
y = 2*x + np.random.randn(100)
scatter_plot(x,y,"X","Y","相关性散点图",save_path="散点图.png")

# --------------------
# 3. 热力图（相关系数）
# --------------------
data = np.random.rand(4,4)
corr = np.corrcoef(data)
heatmap(
    corr,
    ["风速","光照","温度","负荷"],
    ["风速","光照","温度","负荷"],
    title="相关性热力图",
    save_path="热力图.png"
)

# --------------------
# 4. 3D 曲面图
# --------------------
x = np.linspace(-2,2,100)
y = np.linspace(-2,2,100)
X,Y = np.meshgrid(x,y)
Z = X**2 + Y**2

plot_3d_surface(
    x,y,Z,
    xlabel="X",ylabel="Y",zlabel="Z",
    title="3D优化曲面",
    save_path="3d图.png"
)
```
